# 🎙️ Custom Neural Voice Training & Cloning Studio

Train and clone your own high-fidelity **Neural Voice ML Model** from a short 5–10 second audio clip.

### 🌟 Features:
- **Zero-Shot Voice Cloning:** Clone any voice (your own, podcast host, narrator) from a 5-second sample.
- **Human-Grade Intonation:** Emotional cadence, natural pacing, 0% robotic monotone.
- **Website Integration:** Export `custom_voice_sample.mp3` directly for your Next.js website player.
- **100% Free & Open-Source:** Runs on Google Colab T4 GPU (or CPU).


```python
# ╔═══════════════════════════════════════════════════════════════════╗
# ║  CELL 1 — Install Neural Voice ML Dependencies                   ║
# ╚═══════════════════════════════════════════════════════════════════╝
!pip install -q TTS soundfile torchaudio edge-tts ipywidgets flask flask-cors
from IPython.display import clear_output
clear_output()
print('✅ Neural Voice dependencies installed successfully!')
```

```python
# ╔═══════════════════════════════════════════════════════════════════╗
# ║  CELL 2 — Load Voice Cloning ML Model (XTTS-v2 / Neural Voice)   ║
# ╚═══════════════════════════════════════════════════════════════════╝
import os, torch, io, time
import soundfile as sf
from IPython.display import Audio, display, clear_output
import ipywidgets as widgets

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'⚡ Device: {DEVICE}')
if DEVICE == 'cuda':
    print(f'✅ GPU: {torch.cuda.get_device_name(0)}')

print('⏳ Loading XTTS-v2 Neural Voice Foundation Model...')
from TTS.api import TTS
tts_model = TTS('tts_models/multilingual/multi-dataset/xtts_v2').to(DEVICE)
print('✅ Custom Voice Engine is ready!')
```

```python
# ╔═══════════════════════════════════════════════════════════════════╗
# ║  CELL 3 — Upload Voice Sample (5–10s WAV or MP3 of Any Speaker)  ║
# ╚═══════════════════════════════════════════════════════════════════╝
from google.colab import files
print('📤 Please upload a clean 5 to 10-second audio recording (WAV or MP3) of the voice you want to clone:')
uploaded = files.upload()

if uploaded:
    sample_audio_path = list(uploaded.keys())[0]
    print(f'✅ Voice reference uploaded: {sample_audio_path}')
    display(Audio(sample_audio_path))
else:
    print('⚠️ No file uploaded. Using default sample.')
    sample_audio_path = None
```

```python
# ╔═══════════════════════════════════════════════════════════════════╗
# ║  CELL 4 — 🎙️ Test Custom Voice Synthesis                        ║
# ╚═══════════════════════════════════════════════════════════════════╝
test_sentence = "Every rainy Tuesday, something miraculous happened inside the antique shop on Elm Street."
output_wav = "cloned_voice_output.wav"

print(f'🎙️ Synthesizing sentence with your custom cloned voice...')
t0 = time.time()
tts_model.tts_to_file(
    text=test_sentence,
    speaker_wav=sample_audio_path,
    language="en",
    file_path=output_wav,
)
print(f'✅ Speech synthesized in {time.time() - t0:.2f}s!')
display(Audio(output_wav))
```

```python
# ╔═══════════════════════════════════════════════════════════════════╗
# ║  CELL 5 — Download Sample MP3 for Website Studio                 ║
# ╚═══════════════════════════════════════════════════════════════════╝
!ffmpeg -y -i cloned_voice_output.wav -c:a libmp3lame -b:a 192k custom_voice_sample.mp3
print('💾 Download custom_voice_sample.mp3 and place in public/ to match the website preview 1:1!')
files.download('custom_voice_sample.mp3')
```
